## 1. Setup

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing

sns.set_theme(style="whitegrid")
pd.set_option("display.precision", 3)
```

```
housing = fetch_california_housing(as_frame=True)
df = housing.frame

df.shape
```

**Réponse** : 20 640 districts (lignes), 9 colonnes (8 features + 1 target `MedHouseVal`).

**Réponses aux points d'observation** :

### Une première visualisation

```
plt.figure(figsize=(7, 4))
sns.histplot(df["MedHouseVal"], bins=40)
plt.xlabel("Valeur médiane du logement (100k $)")
plt.title("Distribution de la target — MedHouseVal")
plt.show()
```

### Une relation entre deux variables

```
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="MedInc", y="MedHouseVal", alpha=0.2)
plt.xlabel("Revenu médian (10k $)")
plt.ylabel("Valeur médiane du logement (100k $)")
plt.title("Revenu médian vs. Prix médian du logement")
plt.show()
```

**Discussion attendue** : le nuage de points montre une tendance croissante assez nette — plus le revenu médian est élevé, plus le prix médian du logement tend à l'être aussi. Sans ML, on pourrait tracer "à l'œil" une droite qui passe au milieu du nuage. C'est exactement ce que va faire la régression linéaire en Séance 2, mais de façon systématique et optimale plutôt qu'approximative.

## 5. Auto-évaluation — éléments de réponse

1. Une **feature** est une variable d'entrée utilisée pour prédire ; la **target** est la variable qu'on cherche à prédire.
2. En supervisé, on dispose de la target pendant l'entraînement (ex: prix connu) ; en non-supervisé, non (ex: on regroupe des données sans savoir à l'avance les groupes).
3. Parce qu'un modèle peut « mémoriser » les données déjà vues sans avoir réellement appris une règle générale (overfitting) — il faut donc le tester sur des données qu'il n'a jamais rencontrées pour évaluer sa vraie capacité de généralisation.
