### Une première visualisation

Distribution de la variable cible, `MedHouseVal` (valeur médiane du logement, en centaines de milliers de $).

```
plt.figure(figsize=(7, 4))
sns.histplot(df["MedHouseVal"], bins=40)
plt.xlabel("Valeur médiane du logement (100k $)")
plt.title("Distribution de la target — MedHouseVal")
plt.show()
```

### Une relation entre deux variables

On regarde si le revenu médian du foyer (`MedInc`) semble lié au prix du logement.

```
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="MedInc", y="MedHouseVal", alpha=0.2)
plt.xlabel("Revenu médian (10k $)")
plt.ylabel("Valeur médiane du logement (100k $)")
plt.title("Revenu médian vs. Prix médian du logement")
plt.show()
```
